```python
import time, random

def naive_neighbour_count(n, cutoff=1.2, seed=0):
    random.seed(seed)
    pts = [(random.random()*10, random.random()*10, random.random()*10) for _ in range(n)]
    count = 0
    for i in range(n):
        xi, yi, zi = pts[i]
        for j in range(i+1, n):
            xj, yj, zj = pts[j]
            d2 = (xi-xj)**2 + (yi-yj)**2 + (zi-zj)**2
            if d2 < cutoff*cutoff:
                count += 1
    return count

for n in (2000, 4000, 6000):
    t0 = time.time()
    c = naive_neighbour_count(n)
    wall = time.time() - t0
    print(f'n={n} pairs_within_cutoff={c} wall_s={wall:.3f}')
```
